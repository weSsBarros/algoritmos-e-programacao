```python
def coletar_nota(qtd_notas):
    notas = []
    contador = 1
    while contador <= qtd_notas:
        nota =float(input(f"Digite a nota {contador}: "))
        if nota < 0 or nota > 10:
            print("Nota inválida. Digite uma nota entre 0 e 10.")

        else: 
            notas.append(nota)
            contador += 1
    return notas
```

```python
notas = [0.0] * 6

for i in range(6):
    while True:
        try:
            nota = float(input(f"Digite a nota do aluno {i + 1}: "))
        except ValueError:
            print("Valor inválido! Digite um número.")
            continue
        if 0.0 <= nota <= 10.0:
            notas[i] = nota
            break
        print("Nota inválida! Digite um valor entre 0.0 e 10.0.")

print("\nNotas armazenadas:")
for i in range(6):
    print(f"Aluno {i + 1}: {notas[i]:.1f}")
```

```python
soma = 0.0
for i in range(6):
    soma += notas[i]
media = soma / 6

acima = 0
abaixo = []
for i in range(6):
    if notas[i] >= media:
        acima += 1
    else:
        abaixo.append(notas[i])

print(f"Média geral da turma: {media:.2f}")
print(f"Alunos com nota acima ou igual à média: {acima}")
print("Notas abaixo da média:", abaixo)
```

```python
for i in range(6):
    if notas[i] < media:
        while True:
            try:
                nota_recuperacao = float(input(f"Digite a nota de recuperação do aluno {i + 1} (nota atual {notas[i]:.1f}): "))
            except ValueError:
                print("Valor inválido! Digite um número.")
                continue
            if 0.0 <= nota_recuperacao <= 10.0:
                break
            print("Nota inválida! Digite um valor entre 0.0 e 10.0.")
        if nota_recuperacao > notas[i]:
            notas[i] = nota_recuperacao

print("\nNotas atualizadas:")
for i in range(6):
    print(f"Aluno {i + 1}: {notas[i]:.1f}")

soma = 0.0
for i in range(6):
    soma += notas[i]
nova_media = soma / 6
print(f"Nova média geral da turma: {nova_media:.2f}")
```

```python
dias = ["Domingo", "Segunda", "Terça", "Quarta", "Quinta", "Sexta", "Sábado"]
temperaturas = [0.0] * 7

for i in range(7):
    while True:
        try:
            temp = float(input(f"Digite a temperatura máxima de {dias[i]}: "))
        except ValueError:
            print("Valor inválido! Digite um número.")
            continue
        if 10.0 <= temp <= 50.0:
            temperaturas[i] = temp
            break
        print("Temperatura inválida! Digite um valor entre 10.0 e 50.0.")

print("\nTemperaturas registradas:")
for i in range(7):
    print(f"[{i}] {dias[i]}: {temperaturas[i]:.1f}°C")
```

```python
maior = temperaturas[0]
menor = temperaturas[0]
dia_maior = 0
alertas = 0

for i in range(7):
    if temperaturas[i] > maior:
        maior = temperaturas[i]
        dia_maior = i
    if temperaturas[i] < menor:
        menor = temperaturas[i]
    if temperaturas[i] > 32.0:
        alertas += 1

print(f"Maior temperatura: {maior:.1f}°C")
print(f"Menor temperatura: {menor:.1f}°C")
print(f"Dias com Alerta de Calor: {alertas}")
print(f"Dia da maior temperatura: índice {dia_maior} ({dias[dia_maior]})")
```

```python
for i in range(7):
    if temperaturas[i] > 30.0:
        temperaturas[i] -= 4.5
        if temperaturas[i] < 15.0:
            temperaturas[i] = 15.0

print("Temperaturas após a frente fria:")
for i in range(7):
    print(f"{dias[i]}: {temperaturas[i]:.1f}°C")

soma = 0.0
for i in range(7):
    soma += temperaturas[i]
print(f"Nova média da semana: {soma / 7:.2f}°C")
```
